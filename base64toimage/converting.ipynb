```python
import base64
import json
from pathlib import Path


BASE_DIR = Path(r"C:\Projects\projects-ai\gestor_observaciones\base64toimage")

INPUT_JSON = BASE_DIR / "images_base64.json"
OUTPUT_DIR = BASE_DIR / "output_images"


def clean_base64_value(base64_value: str) -> str:
    """
    Soporta Base64 puro o formato data URI:
    data:image/png;base64,...
    """
    if base64_value.startswith("data:") and "," in base64_value:
        return base64_value.split(",", 1)[1]

    return base64_value


def convert_base64_json_to_images(input_json: Path, output_folder: Path):
    if not input_json.exists():
        raise FileNotFoundError(f"No existe el archivo JSON: {input_json}")

    output_folder.mkdir(parents=True, exist_ok=True)

    with open(input_json, "r", encoding="utf-8") as file:
        data = json.load(file)

    if not isinstance(data, list):
        raise ValueError("El JSON debe contener una lista de objetos con filename y base64.")

    restored_count = 0

    for item in data:
        filename = item.get("filename")
        base64_value = item.get("base64")

        if not filename or not base64_value:
            print(f"Registro omitido: falta filename o base64")
            continue

        clean_value = clean_base64_value(base64_value)
        image_bytes = base64.b64decode(clean_value)

        output_file = output_folder / filename

        with open(output_file, "wb") as file:
            file.write(image_bytes)

        restored_count += 1

    print(f"Imágenes restauradas: {restored_count}")
    print(f"Carpeta de salida: {output_folder}")


if __name__ == "__main__":
    convert_base64_json_to_images(INPUT_JSON, OUTPUT_DIR)
```
