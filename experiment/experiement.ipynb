```python
import os
from pathlib import Path
```

```python
KNOWLEDGE_DIR = Path("../knowledge")
KNOWLEDGE_DIR.is_dir()
```

```python
STATIC_CHUNK_CHARS = 1800
STATIC_OVERLAP_CHARS = 200

DYNAMIC_MAX_CHARS = 2200
DYNAMIC_MIN_CHARS = 250

SUPPORTED_EXTENSIONS = {".md"}
```

```python
def load_documents(base_dir: Path, extensions):
    for path in sorted(base_dir.rglob("*")):
        if path.is_file() and path.suffix.lower() in extensions:
            print(path)

load_documents(KNOWLEDGE_DIR, SUPPORTED_EXTENSIONS)
    
```

```python
# use sorted function to return deterministic outputs (the order could varies deppends OS)
for path in sorted(KNOWLEDGE_DIR.rglob("*")):
    if path.is_file() and path.suffix.lower() in SUPPORTED_EXTENSIONS:
        # filter with is_file because rglob returns both files and directories.
        # in this case we want only .md file, that's why wy filter it
        print(path)
```
